```python
import pandas as pd


df = pd.read_csv('/content/WA_Fn-UseC_-Telco-Customer-Churn.csv')


df.head()
```

```python
df.info()
```

```python

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

df = df.dropna()


df = df.drop(columns=['customerID'])

print("Cleaned Data Shape:", df.shape)
```

```python

df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# 'drop_first=True' prevents redundant data (e.g., if 'Male' is 1, 'Female' is automatically 0)
df_encoded = pd.get_dummies(df, drop_first=True)


df_encoded.head()
```

```python

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

X = df_encoded.drop(columns=['Churn'])
y = df_encoded['Churn']


# 'random_state=42' just ensures we get the exact same split if we run it again
X_train, X_test, y_train, y_test = train_test_split(X, y, test_test_split=0.2, random_state=42)


model = RandomForestClassifier(random_state=42)


model.fit(X_train, y_train)


predictions = model.predict(X_test)


accuracy = accuracy_score(y_test, predictions)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv('Lead Scoring.csv')


print(f"Dataset contains {df.shape[0]} leads and {df.shape[1]} columns.\n")

df.head(3)
```

```python

conversion_distribution = df['Converted'].value_counts(normalize=True) * 100

print("--- Lead Conversion Distribution ---")
print(f"Did NOT Convert (0): {conversion_distribution[0]:.2f}%")
print(f"Successfully Converted (1): {conversion_distribution[1]:.2f}%")
```

```python

for col in df.select_dtypes(include='object').columns:
    select_count = (df[col] == 'Select').sum()
    if select_count > 0:
        print(f"Column '{col}' has {select_count} hidden missing values ('Select')")
```

```python

df = df.replace('Select', np.nan)


missing_percentages = (df.isnull().sum() / len(df)) * 100


missing_data_report = missing_percentages[missing_percentages > 0].sort_values(ascending=False)

print("--- Percentage of True Missing Data Per Column ---")
print(round(missing_data_report, 2))
```

```python

columns_to_drop = missing_percentages[missing_percentages > 45].index.tolist()


columns_to_drop.extend(['Prospect ID', 'Lead Number'])

print(f"Dropping the following columns due to high missingness or irrelevance:\n{columns_to_drop}\n")


df_cleaned = df.drop(columns=columns_to_drop)

print(f"New dataset shape after pruning: {df_cleaned.shape}")
```

```python

categorical_cols_to_impute = ['Specialization', 'What is your current occupation', 'Tags', 'City', 'Lead Origin', 'Lead Source']
# Ensure we only look at columns actually present in our cleaned dataframe
categorical_cols_to_impute = [col for col in categorical_cols_to_impute if col in df_cleaned.columns]

for col in categorical_cols_to_impute:
    df_cleaned[col] = df_cleaned[col].fillna('Not Specified')

numeric_cols_to_impute = ['TotalVisits', 'Page Views Per Visit']
for col in numeric_cols_to_impute:
    if col in df_cleaned.columns:
        median_value = df_cleaned[col].median()
        df_cleaned[col] = df_cleaned[col].fillna(median_value)


print(f"Total missing values left in dataset: {df_cleaned.isnull().sum().sum()}")
print(f"Final Data Shape for Modeling: {df_cleaned.shape}")
```

```python
#
categorical_columns = df_cleaned.select_dtypes(include=['object']).columns
df_cleaned[categorical_columns] = df_cleaned[categorical_columns].fillna('Not Specified')


numeric_columns = df_cleaned.select_dtypes(include=[np.number]).columns
for col in numeric_columns:
    if df_cleaned[col].isnull().sum() > 0:
        median_value = df_cleaned[col].median()
        df_cleaned[col] = df_cleaned[col].fillna(median_value)


print(f"Total missing values left in dataset: {df_cleaned.isnull().sum().sum()}")
print(f"Final Data Shape for Modeling: {df_cleaned.shape}")
```

```python
#
df_cleaned['Time_Per_Visit'] = df_cleaned['Total Time Spent on Website'] / (df_cleaned['TotalVisits'] + 1)


source_counts = df_cleaned['Lead Source'].value_counts(normalize=True)


rare_sources = source_counts[source_counts < 0.02].index.tolist()


df_cleaned['Lead Source'] = df_cleaned['Lead Source'].replace(rare_sources, 'Other')


print("--- Updated Lead Source Categories ---")
print(df_cleaned['Lead Source'].value_counts())

print(f"\nNew Dataset Shape after Feature Engineering: {df_cleaned.shape}")
```

```python
from sklearn.model_selection import train_test_split

# 1. Separate features (X) from our target variable (y = 'Converted')
X = df_cleaned.drop(columns=['Converted'])
y = df_cleaned['Converted']

# 2. One-Hot Encode all text columns into 1s and 0s (forcing integer format)
X_encoded = pd.get_dummies(X, drop_first=True, dtype=int)

# 3. Stratified Train-Test Split (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Encoded Features Shape: {X_encoded.shape}")
print(f"Training set size: {X_train.shape[0]} leads")
print(f"Testing set size: {X_test.shape[0]} leads")
```

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score

# 1. Initialize the XGBoost Classifier with enterprise baseline hyperparameters
xgb_model = xgb.XGBClassifier(
    random_state=42,
    learning_rate=0.1,    # Controls how fast the model learns from errors
    n_estimators=100,     # Number of sequential trees to build
    max_depth=5,          # Maximum depth of each tree to prevent overfitting
    eval_metric='logloss'
)

# 2. Train the model
xgb_model.fit(X_train, y_train)

# 3. Generate predictions on unseen test data
y_pred = xgb_model.predict(X_test)
y_proba = xgb_model.predict_proba(X_test)[:, 1] # Probability scores for lead conversion

# 4. Evaluate the metrics
print(f"Overall Model Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("--- Detailed Classification Report ---")
print(classification_report(y_test, y_pred))
```

```python
# 1. Create a clean results DataFrame using our unseen test data
pipeline_results = pd.DataFrame({
    'Actual_Outcome': y_test.values,
    'Conversion_Probability': y_proba  # The raw probability calculated by XGBoost
})

# 2. Scale the probability to a clean 0-100 Lead Score
pipeline_results['Lead_Score'] = (pipeline_results['Conversion_Probability'] * 100).round(1)

# 3. Apply operational tier thresholds based on business logic
def assign_lead_tier(score):
    if score >= 80:
        return 'HOT (Immediate Outbound)'
    elif score >= 35:
        return 'WARM (Nurture Campaign)'
    else:
        return 'COLD (Automated Drop)'

pipeline_results['Operational_Tier'] = pipeline_results['Lead_Score'].apply(assign_lead_tier)

# 4. Display the volume distribution across our new tiers
print("--- Lead Volume Allocation Across Tiers ---")
print(pipeline_results['Operational_Tier'].value_counts())

# 5. Take a look at 5 random records from the newly optimized pipeline
print("\n--- Optimized Salesforce Pipeline Sample ---")
pipeline_results.sample(5, random_state=10)
```

```python
# 1. Extract feature importances from the trained XGBoost model
importances = xgb_model.feature_importances_
feature_names = X_encoded.columns

# 2. Organize them cleanly into a DataFrame
importance_df = pd.DataFrame({
    'Salesforce_Feature': feature_names,
    'Impact_Score': importances
})

# 3. Sort from highest impact to lowest impact
importance_df = importance_df.sort_values(by='Impact_Score', ascending=False)

print("--- Top 10 Structural Drivers of Lead Conversion ---")
importance_df.head(10)
```
